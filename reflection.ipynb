## Visualization Selection
I chose horizontal bar charts as the primary visualization type because the dataset has 33 categories, and horizontal layouts are significantly easier to read than vertical bars when category labels are long. Bar charts were appropriate for most questions since I was comparing aggregated values across discrete groups.
For the Free vs Paid comparison I used side-by-side bar charts to keep the visual clean and make the contrast immediately obvious. The installs comparison is very striking. The median free app gets 500,000 installs versus 5,000 for paid apps, a 100x difference that a bar chart communicates instantly.
For size and price analysis I used binned bar charts rather than scatter plots. This was a deliberate decision driven by a data limitation. As reflected in the dataset, Google pre-buckets install counts into ranges like "10,000+" instead of than reporting exact figures. When plotted on a scatter plot all points to stack in horizontal bands, making the chart misleading. Binning the independent variable and showing average installs per bin produced a much cleaner and more readable visualization.

## Challenges in the Cleaning Process
The dataset had a ton of impurities. The Installs, Price, and Size columns were all stored as strings with characters like commas, plus signs, dollar signs, and unit suffixes that had to be stripped before numeric analysis was possible. The Size column values came in three formats: megabytes ("19M"), kilobytes ("4.2k"), and "Varies with device." The first two were converted to a consistent float in MB and the third was set to null since no meaningful numeric value could be assigned.
The Rating column contained values above 5.0, with a maximum of 19.0 showing up. These rows were filtered out entirely. Additionally 1,474 rows had null ratings and were subsequently dropped rather than imputed because filling them with a mean or median would have corrupted category-level rating comparisons, which were important for my analysis.
Type column contained a value of "0" alongside the valid values "Free" and "Paid." Those rows were removed. 
After all cleaning steps the dataset went from 10,841 rows to 8,890.

## Dashboard Design Decisions
I organized the dashboard into three pages following a logical narrative for a startup stakeholder. Page 1 covers market landscape, how competitive each category is and which ones lead in total installs. Page 2 covers quality and ratings, what makes a well-rated app and how monetization type affects performance. Page 3 covers technical characteristics, how app size and price relate to install success.
I think like this ordering mirrors how a startup would actually approach the decision: first understand the competitive landscape, then understand what quality looks like, then make technical and pricing decisions.
For interactivity I implemented a Top N Categories parameter on allowing users to control how many categories are displayed rather than being overwhelmed by all 33 at once. Category and Type filters apply across all worksheets so viewers can explore specific segments. Navigation buttons connect all three pages with clear labeling although they may be a little off center.
The gradient color scheme represents magnitude within charts. Darker bars represent higher values, and lighter, lower.
One design limitation of note is the Free vs Paid installs chart where the Paid bar is nearly invisible due to the huge difference in median installs. It's accurate but could appear to be a broken chart. 

## Key Recommendations for the Startup
Based on the analysis, a startup entering the Google Play Store should consider launching a free app targeting a broad "Everyone" audience in a mid-competition category like Finance or Productivity rather than the oversaturated Family or Game categories. App size around 50-70MB correlates with higher average installs, suggesting users favor feature-rich experiences without punishing their storage. If monetization is needed, in-app purchases or ads should be considered as opposed to upfront pricing given the install gap between free and paid apps.
